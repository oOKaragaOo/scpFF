```python
import pandas as pd
import glob

# รวมทุกไฟล์ที่ตรง pattern
files = glob.glob("country_airport_stats_*.csv")

# อ่านและรวมเป็น DataFrame เดียว
df = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)

# นับจำนวนแต่ละ country_slug
counts = df['country_slug'].value_counts()

print(counts)
```
